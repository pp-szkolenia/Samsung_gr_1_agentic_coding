## Rodzaje splitterów

### `CharacterTextSplitter`

Dzieli tekst na podstawie określonego separatora (znaku lub ciągu znaków) a także oczekiwanej maksymalnej długości chunka.

```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
splitter = CharacterTextSplitter(
    separator="\n\n",  
    chunk_size=1000,  # max. number of characters
    chunk_overlap=500
)
```

```python
chunks = splitter.split_text(text)
chunks
```

```python
len(chunks)
```

```python
print(chunks[0])
```

```python
len(chunks[0])
```

### `RecursiveCharacterTextSplitter`

Ulepszona wersja `CharacterTextSplitter`. Dzieli tekst iteracyjnie używając kilku separatorów, dzięki czemu lepiej potrafi oddać podział na akapity, przy uwzględnieniu długości chunka.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ". "],
    chunk_size=1000,
    chunk_overlap=500
)
```

```python
chunks = splitter.split_text(text)
chunks
```

```python
len(chunks)
```

```python
print(chunks[0])
```

```python
len(chunks[0])
```

```python
len(chunks[2])
```

### `TokenTextSplitter`

Służy do podziału tekstu na podstawie liczby tokenów a nie znaku podziału.

```python
from langchain_text_splitters import TokenTextSplitter
```

```python
splitter = TokenTextSplitter(
    encoding_name='cl100k_base',
    chunk_size=400,  # number of tokens
    chunk_overlap=200 # number of tokens
)
```

```python
chunks = splitter.split_text(text)
chunks
```

```python
len(chunks)
```

```python
print(chunks[0])
```

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")
tokens = encoding.encode(chunks[0])
len(tokens)
```
