# Zasada działania systemu RAG

**RAG** – Retrieval-Augmented Generation

**Cel**: Umożliwienie modelowi generacji w oparciu o informacje zawarte w dostarczonych mu dokumentach.

## Koncepcja embeddingu

Embeddingi to słowa lub dokumenty (czyli ciągi tekstu) zamienione na wektory na podstawie ich znaczenia. Dokumenty o podobnym znaczeniu będą blisko siebie w przestrzeni wektorowej a te niezwiązane tematycznie – daleko.
