### Pliki .pdf

Istnieje wiele parserów plików `.pdf` w LangChain. Jednym z nich jest `PyMuPDFLoader`, który obsługuje między innymi metadane.

```python
from langchain_community.document_loaders import PyMuPDFLoader
```

**Single page**

```python
loader = PyMuPDFLoader("documents/AI.pdf")
loader
```

```python
documents = loader.load()
documents
```

```python
documents[0].page_content
```

```python
documents[0].metadata
```

**Multi page**

`PyMuPDFLoader` automatycznie dzieli plik pdf na dokumenty na podstawie stron.

```python
loader = PyMuPDFLoader("documents/AI_long.pdf")
documents = loader.load()
```

```python
len(documents)
```

```python
for doc in documents:
    print(doc.metadata, "\n")
```

```python
content = "\n\n=====\n\n".join([doc.page_content for doc in documents])

print(content)
```
